```python
# ==========================================
# COMPLETE PIPELINE: PITT DEMENTIA CLASSIFICATION (OPTIMIZED)
# ==========================================
# This code runs everything from data loading to final ensemble results
# Optimized for speed with reduced grid search and folds
# ==========================================

import os
import re
import zipfile
import warnings
import numpy as np
import pandas as pd
from tqdm import tqdm
warnings.filterwarnings('ignore')

# ==========================================
# PART 1: MOUNT DRIVE AND LOAD DATA
# ==========================================

from google.colab import drive
drive.mount('/content/drive')

print("\n" + "="*70)
print("📁 LOADING DATA")
print("="*70)

# ------------------------------------------
# STEP 1: LOAD AUDIO FILES
# ------------------------------------------

base_dir = "/content/drive/MyDrive/DementiaBank_Audio"
pitt_data = []

for class_name in ['Dementia', 'Control']:
    folder = os.path.join(base_dir, class_name)
    if os.path.exists(folder):
        for file in os.listdir(folder):
            if file.endswith('.mp3'):
                fp = os.path.join(folder, file)
                if os.path.getsize(fp) > 10000:
                    pid = file.replace('.mp3', '')
                    dx = 'ad' if class_name == 'Dementia' else 'cn'
                    pitt_data.append({'participant_id': pid, 'dx': dx, 'audio_path': fp})

pitt_df = pd.DataFrame(pitt_data)

if pitt_df['participant_id'].duplicated().sum():
    pitt_df = pitt_df.drop_duplicates(subset='participant_id', keep='first')

def get_subject_id(pid):
    m = re.match(r'^(.*)-\d+$', pid)
    return m.group(1) if m else pid

pitt_df['subject_id'] = pitt_df['participant_id'].apply(get_subject_id)

print(f"✅ Audio files loaded: {len(pitt_df)} recordings")
print(f"   AD: {(pitt_df['dx']=='ad').sum()}, CN: {(pitt_df['dx']=='cn').sum()}")
print(f"   Unique subjects: {pitt_df['subject_id'].nunique()}")

# ------------------------------------------
# STEP 2: PARSE TRANSCRIPTS
# ------------------------------------------

print("\n📄 Loading transcripts...")

def parse_cha_file(filepath):
    try:
        with open(filepath, 'r', encoding='utf-8', errors='ignore') as f:
            lines = f.readlines()
        par_lines = []
        for line in lines:
            if line.startswith('*PAR:') or line.startswith('*PAR\t'):
                text = line.split('\t', 1)[1].strip() if '\t' in line else \
                       line.replace('*PAR:', '').strip()
                text = re.sub(r'&\*[A-Z]+:\S*', '', text)
                text = re.sub(r'\[.*?\]', '', text)
                text = re.sub(r'\(\.+\)', '', text)
                text = re.sub(r'\d+_\d+', '', text)
                text = re.sub(r'&[\w-]+', '', text)
                text = re.sub(r'[<>]', '', text)
                text = re.sub(r'\s+', ' ', text).strip()
                if text and len(text) > 3:
                    par_lines.append(text)
        return ' '.join(par_lines)
    except Exception:
        return ''

# Find transcripts
zip_path = "/content/drive/MyDrive/DementiaBank_Audio/Transcripts/Pitt.zip"
extract_path = "/content/pitt_transcripts"

if os.path.exists(zip_path):
    if not os.path.exists(extract_path):
        print("   Extracting transcripts...")
        with zipfile.ZipFile(zip_path, 'r') as z:
            z.extractall(extract_path)
        print("   ✅ Extraction complete")
    else:
        print("   Using existing extraction")

# Find .cha files
all_cha_files = []
if os.path.exists(extract_path):
    for root, dirs, files in os.walk(extract_path):
        for file in files:
            if file.endswith('.cha'):
                all_cha_files.append(os.path.join(root, file))

print(f"   Found {len(all_cha_files)} .cha files")

# Filter for Cookie Theft
cha_files = [f for f in all_cha_files if 'cookie' in f.lower()]
print(f"   Cookie Theft files: {len(cha_files)}")

# Parse transcripts
transcript_by_pid = {}
for cha_path in tqdm(cha_files, desc="Parsing"):
    pid = os.path.basename(cha_path).replace('.cha', '')
    text = parse_cha_file(cha_path)
    if text and len(text) > 10:
        transcript_by_pid[pid] = text

pitt_transcripts_df = pd.DataFrame([
    {'participant_id': pid, 'transcript': txt} for pid, txt in transcript_by_pid.items()
])

print(f"✅ Transcripts parsed: {len(pitt_transcripts_df)} participants")

# ------------------------------------------
# STEP 3: MERGE AUDIO + TRANSCRIPT
# ------------------------------------------

print("\n🔗 Merging data...")

df = pitt_df.merge(pitt_transcripts_df, on='participant_id', how='inner')
df = df[df['transcript'].notna()].reset_index(drop=True)

print(f"✅ Final dataset: {len(df)} participants")
print(f"   AD: {(df['dx']=='ad').sum()}, CN: {(df['dx']=='cn').sum()}")
print(f"   Unique subjects: {df['subject_id'].nunique()}")

# ==========================================
# PART 2: INSTALL DEPENDENCIES
# ==========================================

print("\n" + "="*70)
print("📦 INSTALLING DEPENDENCIES")
print("="*70)

!pip -q install praat-parselmouth spacy librosa soundfile lexicalrichness nltk shap upsetplot imbalanced-learn lightgbm catboost
!python -m spacy download en_core_web_sm

import parselmouth
from parselmouth.praat import call
import spacy
from lexicalrichness import LexicalRichness
from sklearn.model_selection import StratifiedGroupKFold, GridSearchCV, RandomizedSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, VotingClassifier
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.naive_bayes import GaussianNB
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.metrics import accuracy_score, roc_auc_score, f1_score
from sklearn.feature_selection import SelectKBest, f_classif
from imblearn.over_sampling import SMOTE
from sklearn.base import clone
from collections import Counter
from scipy import stats
import time

nlp = spacy.load("en_core_web_sm")

# ==========================================
# PART 3: FEATURE EXTRACTION
# ==========================================

print("\n" + "="*70)
print("🎵 EXTRACTING FEATURES")
print("="*70)

# ------------------------------------------
# Acoustic Features
# ------------------------------------------

def extract_acoustic_features(file_path, silence_threshold=-25, min_silence_duration=0.1):
    try:
        sound = parselmouth.Sound(file_path)
        duration = sound.duration
        textgrid = call(sound, "To TextGrid (silences)", 100, 0, silence_threshold,
                         min_silence_duration, min_silence_duration, "silent", "sounding")
        n_intervals = call(textgrid, "Get number of intervals", 1)
        pause_durations, speech_durations = [], []
        for i in range(1, n_intervals + 1):
            label = call(textgrid, "Get label of interval", 1, i)
            start = call(textgrid, "Get start point", 1, i)
            end = call(textgrid, "Get end point", 1, i)
            (pause_durations if label == "silent" else speech_durations).append(end - start)
        if not pause_durations: pause_durations = [0]
        if not speech_durations: speech_durations = [0]
        total_pause = np.sum(pause_durations)
        total_speech = np.sum(speech_durations)
        return {"duration_sec": duration, "pause_count": len(pause_durations),
                "mean_pause_duration": np.mean(pause_durations), "max_pause_duration": np.max(pause_durations),
                "total_pause_time": total_pause, "silence_ratio": total_pause / duration if duration > 0 else 0,
                "speech_duration": total_speech, "mean_speech_segment": np.mean(speech_durations),
                "max_speech_segment": np.max(speech_durations)}
    except Exception:
        return {k: 0 for k in ["duration_sec","pause_count","mean_pause_duration","max_pause_duration",
                                "total_pause_time","silence_ratio","speech_duration","mean_speech_segment","max_speech_segment"]}

print("\n🎵 Extracting acoustic features...")
acoustic_features = []
for _, row in tqdm(df.iterrows(), total=len(df)):
    feats = extract_acoustic_features(row["audio_path"])
    feats["participant_id"] = row["participant_id"]
    acoustic_features.append(feats)

acoustic_df = pd.DataFrame(acoustic_features)
df = df.merge(acoustic_df, on="participant_id", how="left")

# ------------------------------------------
# Linguistic Features
# ------------------------------------------

def extract_linguistic_features(text, duration_sec):
    doc = nlp(str(text))
    words = [t for t in doc if t.is_alpha]
    total_words = len(words)
    if total_words == 0:
        return {k: 0 for k in ["speech_rate_wpm","word_count","filled_pauses","type_token_ratio","mattr","mtld",
                                "pronoun_ratio","noun_ratio","verb_ratio","lexical_density","avg_sentence_length",
                                "avg_word_length","lexical_sophistication"]}
    lr = LexicalRichness(" ".join([t.text for t in words]))
    pronouns = sum(t.pos_ == "PRON" for t in words)
    nouns = sum(t.pos_ == "NOUN" for t in words)
    verbs = sum(t.pos_ == "VERB" for t in words)
    adjectives = sum(t.pos_ == "ADJ" for t in words)
    adverbs = sum(t.pos_ == "ADV" for t in words)
    filled = len(re.findall(r"\b(um|uh|erm|ah)\b", str(text).lower()))
    sentences = list(doc.sents)
    avg_sent_len = total_words / len(sentences) if len(sentences) else total_words
    content_words = nouns + verbs + adjectives + adverbs
    lexical_density = content_words / total_words
    avg_word_len = np.mean([len(t.text) for t in words])
    long_words = sum(len(t.text) >= 6 for t in words)
    lex_soph = long_words / total_words
    speech_rate = total_words / (duration_sec / 60) if duration_sec > 0 else 0
    try: mattr = lr.mattr(window_size=25)
    except Exception: mattr = 0
    try: mtld = lr.mtld(threshold=0.72)
    except Exception: mtld = 0
    return {"speech_rate_wpm": speech_rate, "word_count": total_words, "filled_pauses": filled,
            "type_token_ratio": lr.ttr, "mattr": mattr, "mtld": mtld, "pronoun_ratio": pronouns/total_words,
            "noun_ratio": nouns/total_words, "verb_ratio": verbs/total_words, "lexical_density": lexical_density,
            "avg_sentence_length": avg_sent_len, "avg_word_length": avg_word_len, "lexical_sophistication": lex_soph}

print("\n🔤 Extracting linguistic features...")
linguistic_features = []
for _, row in tqdm(df.iterrows(), total=len(df)):
    feats = extract_linguistic_features(row["transcript"], row["duration_sec"])
    feats["participant_id"] = row["participant_id"]
    linguistic_features.append(feats)

ling_df = pd.DataFrame(linguistic_features)
df = df.merge(ling_df, on="participant_id", how="left")

# ------------------------------------------
# Define All Features
# ------------------------------------------

all_features = ["duration_sec", "pause_count", "mean_pause_duration", "max_pause_duration",
                 "total_pause_time", "silence_ratio", "speech_duration", "mean_speech_segment",
                 "max_speech_segment", "speech_rate_wpm", "word_count", "filled_pauses",
                 "type_token_ratio", "mattr", "mtld", "pronoun_ratio", "noun_ratio", "verb_ratio",
                 "lexical_density", "avg_sentence_length", "avg_word_length", "lexical_sophistication"]

df[all_features] = df[all_features].fillna(0)

print(f"\n✅ Feature extraction complete: {df.shape[1]} columns")

# ==========================================
# PART 4: TRAIN/TEST SPLIT
# ==========================================

print("\n" + "="*70)
print("📊 TRAIN/TEST SPLIT")
print("="*70)

from sklearn.model_selection import GroupShuffleSplit

X_full = df[all_features]
y_full = df['dx'].map({'cn': 0, 'ad': 1})
groups_full = df['subject_id']

gss = GroupShuffleSplit(n_splits=1, test_size=0.20, random_state=42)
dev_idx, holdout_idx = next(gss.split(X_full, y_full, groups=groups_full))

X_dev_raw = X_full.iloc[dev_idx]
y_dev = y_full.iloc[dev_idx]
groups_dev = groups_full.iloc[dev_idx]
X_holdout_raw = X_full.iloc[holdout_idx]
y_holdout = y_full.iloc[holdout_idx]

print(f"✅ Dev set: {len(X_dev_raw)} participants / {groups_dev.nunique()} subjects")
print(f"✅ Holdout set: {len(X_holdout_raw)} participants / {groups_full.iloc[holdout_idx].nunique()} subjects")

# ==========================================
# PART 5: FEATURE SELECTION
# ==========================================

print("\n" + "="*70)
print("📋 FEATURE SELECTION")
print("="*70)

corr_matrix = X_dev_raw[all_features].corr().abs()
avg_corr = corr_matrix.mean()

to_drop = set()
for i in range(len(all_features)):
    for j in range(i+1, len(all_features)):
        f1, f2 = all_features[i], all_features[j]
        if f1 in to_drop or f2 in to_drop:
            continue
        if corr_matrix.loc[f1, f2] > 0.85:
            drop = f1 if avg_corr[f1] > avg_corr[f2] else f2
            to_drop.add(drop)
            print(f"   Dropping '{drop}' (r={corr_matrix.loc[f1,f2]:.2f} with '{f2 if drop==f1 else f1}')")

selected_features = [f for f in all_features if f not in to_drop]
print(f"\n✅ Selected {len(selected_features)}/{len(all_features)} features")

X_dev = X_dev_raw[selected_features]
X_holdout = X_holdout_raw[selected_features]

# ==========================================
# PART 6: FEATURE ENGINEERING & SMOTE
# ==========================================

print("\n" + "="*70)
print("🔧 FEATURE ENGINEERING & SMOTE")
print("="*70)

def add_interaction_features(X):
    X_new = X.copy()
    key_features = ['speech_rate_wpm', 'silence_ratio', 'mtld', 'avg_sentence_length']
    for feat in key_features:
        if feat in X.columns:
            X_new[f'{feat}_sq'] = X[feat] ** 2
            X_new[f'{feat}_log'] = np.log1p(X[feat] + 1)

    if 'speech_rate_wpm' in X.columns and 'total_pause_time' in X.columns:
        X_new['rate_pause_interaction'] = X['speech_rate_wpm'] * X['total_pause_time']

    if 'speech_rate_wpm' in X.columns and 'silence_ratio' in X.columns:
        X_new['rate_silence_interaction'] = X['speech_rate_wpm'] * X['silence_ratio']

    return X_new

X_dev_enhanced = add_interaction_features(X_dev)
print(f"   Features after engineering: {X_dev_enhanced.shape[1]}")

# Advanced feature selection
selector = SelectKBest(f_classif, k=min(15, X_dev_enhanced.shape[1]))
selector.fit(X_dev_enhanced, y_dev)
selected_indices = selector.get_support(indices=True)
X_dev_selected = X_dev_enhanced.iloc[:, selected_indices]
selected_feat_names = X_dev_enhanced.columns[selected_indices].tolist()

print(f"   Final features: {len(selected_feat_names)}")

# ------------------------------------------
# SMOTE for class balance
# ------------------------------------------

print("\n⚖️ Applying SMOTE...")
smote = SMOTE(random_state=42, k_neighbors=3)
X_balanced, y_balanced = smote.fit_resample(X_dev_selected, y_dev)

# Create balanced groups for SMOTE samples
original_indices = X_dev_selected.index
original_groups = groups_dev.loc[original_indices].values

n_original = len(X_dev_selected)
n_synthetic = len(X_balanced) - n_original

# Convert string groups to numeric
unique_groups = np.unique(original_groups)
group_to_numeric = {g: i for i, g in enumerate(unique_groups)}
numeric_groups = np.array([group_to_numeric[g] for g in original_groups])

max_numeric_id = max(numeric_groups)
balanced_groups_numeric = np.concatenate([
    numeric_groups,
    np.arange(max_numeric_id + 1, max_numeric_id + 1 + n_synthetic)
])

balanced_groups = balanced_groups_numeric

print(f"   After SMOTE: {len(X_balanced)} samples (AD: {sum(y_balanced==1)}, CN: {sum(y_balanced==0)})")
print(f"   Original samples: {n_original}, Synthetic samples: {n_synthetic}")

# ==========================================
# PART 7: HYPERPARAMETER GRIDS (OPTIMIZED)
# ==========================================

param_grids = {
    'LogisticRegression': {
        'classifier__C': [0.1, 1.0, 10.0],
        'classifier__solver': ['liblinear', 'lbfgs'],
        'classifier__class_weight': ['balanced', None]
    },
    'RandomForest': {
        'n_estimators': [100, 200],
        'max_depth': [10, 20],
        'min_samples_split': [2, 5],
        'min_samples_leaf': [1, 2],
        'class_weight': ['balanced', 'balanced_subsample']
    },
    'XGBoost': {
        'n_estimators': [100, 200],
        'max_depth': [3, 5],
        'learning_rate': [0.05, 0.1],
        'subsample': [0.8, 1.0],
        'colsample_bytree': [0.8, 1.0],
        'min_child_weight': [1, 3],
        'scale_pos_weight': [1, 3]
    },
    'LightGBM': {
        'n_estimators': [100, 200],
        'max_depth': [5, -1],
        'learning_rate': [0.05, 0.1],
        'num_leaves': [31, 63],
        'subsample': [0.8, 1.0],
        'class_weight': ['balanced', None]
    },
    'SVM': {
        'classifier__C': [0.1, 1.0, 10.0],
        'classifier__kernel': ['linear', 'rbf'],
        'classifier__gamma': ['scale', 'auto'],
        'classifier__class_weight': ['balanced', None]
    },
    'MLP': {
        'classifier__hidden_layer_sizes': [(64,), (128,)],
        'classifier__activation': ['relu'],
        'classifier__alpha': [0.0001, 0.001],
        'classifier__learning_rate_init': [0.001],
        'classifier__batch_size': [32]
    },
    'GradientBoosting': {
        'n_estimators': [100, 200],
        'learning_rate': [0.05, 0.1],
        'max_depth': [3, 5],
        'min_samples_split': [2, 5],
        'subsample': [0.8, 1.0]
    }
}

# ==========================================
# PART 8: BASE MODELS (OPTIMIZED)
# ==========================================

base_models = {
    'LogisticRegression': Pipeline([
        ("scaler", RobustScaler()),
        ("classifier", LogisticRegression(max_iter=1000, random_state=42))
    ]),
    'RandomForest': RandomForestClassifier(random_state=42, n_jobs=-1, class_weight='balanced', n_estimators=200),
    'XGBoost': XGBClassifier(
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        scale_pos_weight=len(y_dev[y_dev==0])/len(y_dev[y_dev==1]),
        use_label_encoder=False,
        n_estimators=200,
        max_depth=5,
        learning_rate=0.1
    ),
    'LightGBM': LGBMClassifier(random_state=42, n_jobs=-1, class_weight='balanced', verbose=-1),
    'SVM': Pipeline([
        ("scaler", RobustScaler()),
        ("classifier", SVC(probability=True, random_state=42, class_weight='balanced'))
    ]),
    'MLP': Pipeline([
        ("scaler", RobustScaler()),
        ("classifier", MLPClassifier(max_iter=1500, random_state=42, early_stopping=True))
    ]),
    'GradientBoosting': GradientBoostingClassifier(random_state=42, n_estimators=200)
}

# ==========================================
# PART 9: TRAIN MODELS (OPTIMIZED)
# ==========================================

print("\n" + "="*70)
print("🚀 TRAINING MODELS")
print("="*70)

cv = StratifiedGroupKFold(n_splits=3, shuffle=True, random_state=42)
inner_cv = StratifiedGroupKFold(n_splits=2, shuffle=True, random_state=42)

accuracies = {}
aucs = {}
tuned_models = {}
training_times = {}

for model_name, base_model in base_models.items():
    print(f"\n📊 {model_name}")
    start_time = time.time()

    fold_accuracies = []
    fold_aucs = []

    for fold, (train_idx, test_idx) in enumerate(cv.split(X_balanced, y_balanced, groups=balanced_groups), 1):
        print(f"  Fold {fold}/{cv.n_splits}...", end=" ")

        X_train = X_balanced.iloc[train_idx]
        X_test = X_balanced.iloc[test_idx]
        y_train = y_balanced.iloc[train_idx]
        y_test = y_balanced.iloc[test_idx]
        groups_train = balanced_groups[train_idx]

        if model_name in param_grids:
            # Use RandomizedSearchCV for speed
            grid_search = RandomizedSearchCV(
                estimator=clone(base_model),
                param_distributions=param_grids[model_name],
                n_iter=6,
                cv=inner_cv,
                scoring='accuracy',
                n_jobs=-1,
                verbose=0,
                random_state=42
            )
            grid_search.fit(X_train, y_train, groups=groups_train)
            model = grid_search.best_estimator_
            print(f"done (best: {grid_search.best_score_:.3f})")
        else:
            model = clone(base_model)
            model.fit(X_train, y_train)
            print("done")

        # XGBoost early stopping
        if model_name == 'XGBoost' and hasattr(model, 'fit'):
            try:
                model.fit(X_train, y_train, eval_set=[(X_test, y_test)], early_stopping_rounds=10, verbose=False)
            except:
                pass

        pred = model.predict(X_test)
        proba = model.predict_proba(X_test)[:, 1]

        fold_accuracies.append(accuracy_score(y_test, pred))
        fold_aucs.append(roc_auc_score(y_test, proba))

    # Train final model
    final_model = clone(base_model)
    if model_name in param_grids:
        try:
            final_model.set_params(**grid_search.best_params_)
        except:
            pass
    final_model.fit(X_balanced, y_balanced)
    tuned_models[model_name] = final_model

    accuracies[model_name] = fold_accuracies
    aucs[model_name] = fold_aucs
    training_times[model_name] = time.time() - start_time

    print(f"  ✅ Mean Accuracy: {np.mean(fold_accuracies):.3f} ± {np.std(fold_accuracies):.3f}")
    print(f"  ✅ Mean AUC: {np.mean(fold_aucs):.3f} ± {np.std(fold_aucs):.3f}")
    print(f"  ⏱️  Time: {training_times[model_name]:.1f}s")

# ==========================================
# PART 10: SELECT BEST MODELS & ENSEMBLE
# ==========================================

print("\n" + "="*70)
print("📊 MODEL PERFORMANCE SUMMARY")
print("="*70)

model_performance = {}
for name in base_models.keys():
    if name in accuracies:
        model_performance[name] = np.mean(accuracies[name])

sorted_models = sorted(model_performance.items(), key=lambda x: x[1], reverse=True)

print("\nIndividual Model Accuracies:")
for i, (name, acc) in enumerate(sorted_models, 1):
    print(f"  {i}. {name}: {acc:.3f}")

best_model_name = sorted_models[0][0]
best_accuracy = sorted_models[0][1]
print(f"\n🏆 Best Individual Model: {best_model_name} ({best_accuracy:.3f})")

# Select top models
threshold = 0.69
selected_models = [name for name, acc in sorted_models if acc >= threshold]

if len(selected_models) < 2:
    selected_models = [name for name, acc in sorted_models[:3]]

print(f"\n🔗 Selected models for ensemble:")
for name in selected_models:
    print(f"  - {name}: {model_performance[name]:.3f}")

# Build ensemble
estimators = []
weights = []
for name in selected_models:
    estimators.append((name, tuned_models[name]))
    weights.append(model_performance[name])

voting_clf = VotingClassifier(
    estimators=estimators,
    voting='soft',
    weights=weights
)

# Evaluate ensemble
print("\n📊 Evaluating Ensemble...")
ensemble_accuracies = []
ensemble_aucs = []

for train_idx, test_idx in cv.split(X_balanced, y_balanced, groups=balanced_groups):
    X_train = X_balanced.iloc[train_idx]
    X_test = X_balanced.iloc[test_idx]
    y_train = y_balanced.iloc[train_idx]
    y_test = y_balanced.iloc[test_idx]

    clf = clone(voting_clf)
    clf.fit(X_train, y_train)

    pred = clf.predict(X_test)
    proba = clf.predict_proba(X_test)[:, 1]

    ensemble_accuracies.append(accuracy_score(y_test, pred))
    ensemble_aucs.append(roc_auc_score(y_test, proba))

print(f"\n✅ Ensemble Results:")
print(f"  Accuracy: {np.mean(ensemble_accuracies):.3f} ± {np.std(ensemble_accuracies):.3f}")
print(f"  AUC: {np.mean(ensemble_aucs):.3f} ± {np.std(ensemble_aucs):.3f}")

# Train final ensemble
voting_clf.fit(X_balanced, y_balanced)

# ==========================================
# PART 11: FINAL RESULTS
# ==========================================

print("\n" + "="*70)
print("📊 FINAL COMPARISON")
print("="*70)

comparison_data = []
for name, acc in sorted_models[:5]:
    comparison_data.append({'Model': name, 'Accuracy': f"{acc:.3f}"})

comparison_data.append({'Model': f'Ensemble ({len(selected_models)} models)',
                       'Accuracy': f"{np.mean(ensemble_accuracies):.3f}"})

comparison_df = pd.DataFrame(comparison_data)
print(comparison_df.to_string(index=False))

improvement = np.mean(ensemble_accuracies) - best_accuracy
print(f"\n📈 Improvement: {improvement:.3f} ({improvement*100:.1f}%)")

# ==========================================
# PART 12: EVALUATE ON HOLDOUT SET
# ==========================================

print("\n" + "="*70)
print("🎯 HOLDOUT SET EVALUATION")
print("="*70)

# Apply same feature engineering to holdout
X_holdout_enhanced = add_interaction_features(X_holdout)
X_holdout_selected = X_holdout_enhanced[selected_feat_names]

# Make predictions
holdout_predictions = voting_clf.predict(X_holdout_selected)
holdout_proba = voting_clf.predict_proba(X_holdout_selected)[:, 1]

holdout_accuracy = accuracy_score(y_holdout, holdout_predictions)
holdout_auc = roc_auc_score(y_holdout, holdout_proba)
holdout_f1 = f1_score(y_holdout, holdout_predictions)

print(f"✅ Holdout Set Results:")
print(f"  Accuracy: {holdout_accuracy:.3f}")
print(f"  AUC: {holdout_auc:.3f}")
print(f"  F1 Score: {holdout_f1:.3f}")

# ==========================================
# PART 13: SAVE RESULTS
# ==========================================

print("\n💾 Saving results...")

# Save model performance
perf_df = pd.DataFrame([
    {'Model': name, 'Accuracy': np.mean(accuracies[name]),
     'AUC': np.mean(aucs[name]),
     'Time': training_times[name]} for name in base_models.keys() if name in accuracies
])
perf_df.to_csv('/content/model_performance.csv', index=False)

# Save ensemble predictions on holdout
holdout_df = pd.DataFrame({
    'participant_id': X_holdout.index,
    'true_label': y_holdout,
    'predicted_label': holdout_predictions,
    'probability': holdout_proba
})
holdout_df.to_csv('/content/holdout_predictions.csv', index=False)

# Save feature importance if available
feature_importance_df = pd.DataFrame()
for name in selected_models:
    model = tuned_models[name]
    if hasattr(model, 'feature_importances_'):
        fi = pd.DataFrame({
            'feature': selected_feat_names,
            'importance': model.feature_importances_
        }).sort_values('importance', ascending=False)
        fi.to_csv(f'/content/feature_importance_{name}.csv', index=False)

print("   ✅ Results saved to /content/")
print("      - model_performance.csv")
print("      - holdout_predictions.csv")
print("      - feature_importance_*.csv")

# ==========================================
# PART 14: FINAL SUMMARY
# ==========================================

print("\n" + "="*70)
print("✅ COMPLETE PIPELINE COMPLETED")
print("="*70)
print(f"""
Summary:
---------
Total Samples: {len(df)}
Features Extracted: {len(all_features)}
Features After Selection: {len(selected_features)}
SMOTE Applied: {len(X_balanced)} samples

Best Individual Model: {best_model_name} ({best_accuracy:.3f})
Ensemble Accuracy (CV): {np.mean(ensemble_accuracies):.3f}
Improvement: {improvement:.3f} ({improvement*100:.1f}%)

Holdout Performance:
  Accuracy: {holdout_accuracy:.3f}
  AUC: {holdout_auc:.3f}
  F1: {holdout_f1:.3f}

Training Times:
""")
for name, time_taken in training_times.items():
    print(f"  {name}: {time_taken:.1f}s")

print(f"""
Files Saved:
  - model_performance.csv
  - holdout_predictions.csv
  - feature_importance_*.csv
""")
print("="*70)
```
